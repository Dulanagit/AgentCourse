# And now - TOOLS!

Let's start with a function...

````python
def record_email_tool(email):
    print(f"Tool called to record an email: {email}")
    with open("emails.txt", "a", encoding="utf-8") as f:
        f.write(email + "\n")
    return "Email received"
````

````python
record_email_tool("test@testy.com")
````

## Step 1 - write some json to describe the tool


````python
record_email_tool_json = {
    "name": "record_email_tool",
    "description": "Use this tool to record that a user provided their email address",
    "parameters": {
        "type": "object",
        "properties": {
            "email": {"type": "string", "description": "The email address of this user"}
        },
        "required": ["email"],
        "additionalProperties": False
    }
}
````

````python
tools = [{"type": "function", "function": record_email_tool_json}]
````

````python
tools
````

## Step 2 - a new chat() function

This is where we implement the tool call.

The reality is, it's a bit clunky. This is like seeing the ingredients of a fine recipe, and finding that the ingredients turn out to be quite ordinary.

Tool calling is an "if" statement. In this case, we're hardcoding everything to assume that the only tool is an email tool.

SIDENOTE: If you're thinking - but wait! I should be remembering this so I can do it myself! Then the key point is: this is what Agent Frameworks take care of for you. In practice, you'll likely never type this again yourself. We are shielded from these if statements by the Agent Framework. That's why they're often described as "abstraction layers".

````python
def chat(message, history):
    history = [{"role": h["role"], "content":h["content"]} for h in history]
    messages = [{"role": "system", "content": system_prompt}] + history + [{"role": "user", "content": message}]
    response = openai.chat.completions.create(model=model, messages=messages, tools=tools)
         
    if response.choices[0].finish_reason=="tool_calls":
            message = response.choices[0].message
            print(f"message: {message}")
            tool_call = message.tool_calls[0]
            print(f"tool_call: {tool_call}")
            email = json.loads(tool_call.function.arguments).get("email") #json.loads converts json to dic
            record_email_tool(email)
            messages.append(message)
            messages.append({"role": "tool", "content": "Email recorded", "tool_call_id": tool_call.id})
            response = openai.chat.completions.create(model=model, messages=messages, tools=tools)
            
    return response.choices[0].message.content
````

````python
gr.ChatInterface(chat).launch(inbrowser=True)
````

## Step 3

Our first ever Agent Loop, done without an Agent Framework!

Changes:
1. Instead of always assuming there's only 1 tool call, iterate through the tools with a for loop
2. Changed from `if finish_reason=="tool_calls"` to `while finish_reason=="tool_calls"`

````python
def chat(message, history):
    history = [{"role": h["role"], "content": h["content"]} for h in history]
    messages = [{"role": "system", "content": system_prompt}] + history + [{"role": "user", "content": message}]
    response = openai.chat.completions.create(model=model, messages=messages, tools=tools)
         
    while response.choices[0].finish_reason=="tool_calls":
            message = response.choices[0].message
            messages.append(message)
            for tool_call in message.tool_calls:
                email = json.loads(tool_call.function.arguments).get("email")
                record_email_tool(email)
                messages.append({"role": "tool", "content": "Email recorded", "tool_call_id": tool_call.id})
            response = openai.chat.completions.create(model=model, messages=messages, tools=tools)
            
    return response.choices[0].message.content
````

````python
gr.ChatInterface(chat).launch(inbrowser=True)
````

````python
evaluator_system_prompt = f"""
    # Your role

    You are the evaluator of the responses of a digital twin running on a website, chatting with visitors of the website.
    You represent the person who evaluates the response of the digital twin and ensure it answers only work related questions.
    

    Here are the details of the person that digital twin is representing:

    {summary}

    
    # Context

    Here is a summary of the person's LinkedIn profile so that you can review the relatedness:

    {linkedin} 

    # Rules

    Make sure that the digital twin has responded within the following rules.
    -Engage with the user. Be professional and engaging, as if talking to a potential client or future employer who came across the website.
    -Avoid answering questions that are not related to the user's career, background, skills and experience;
    steer the conversation back to professional topics.

    Always stay in character as the evaluator of the digital twin. You are the evaluator.

    IMPORTANT: If the digital twin has answered something not related to user's career, background, skills and experience or anything not professional and related to private life,give a message to guide 
    the digital twin to answer within it's rules. Else, say it, "The response is approved, you can send the original message to the user". You should always call the evaluation pass tool along with the response.

    

"""
````

````python
evaluation_pass_json = {
    "name": "evaluation_pass",
        "description": "Use this tool to state if the the response is approved or not.",
        "parameters": {
            "type": "object",
            "properties": {
                "is_pass": {"type": "boolean", "description": "true/false value as the approval or not respectively."},
                "feedback_message": {
                "type": "string", 
                "description": "If is_pass is false, provide a message to guide the digital twin. If true, say 'The response is approved'."
                },
            },
            "required": ["is_pass"],
            "additionalProperties": False
        }
}
````

````python
evaluator_tools = [{"type": "function", "function": evaluation_pass_json}]

evaluator_tool_choice = {"type": "function", "function": {"name": "evaluation_pass"}}
````

````python
def evaluate(agent_response):
    messages = [{"role": "system", "content": evaluator_system_prompt}] + [{"role": "user", "content": agent_response}]
    response = openai.chat.completions.create(model=evaluator_model, messages=messages, tools=evaluator_tools, tool_choice=evaluator_tool_choice)

    tool_call = response.choices[0].message.tool_calls[0]
    is_pass = json.loads(tool_call.function.arguments).get("is_pass")
    evaluator_response = json.loads(tool_call.function.arguments).get("feedback_message")

    return is_pass, evaluator_response

````

````python
def chat(message, history):
    evaluation_passed = False
    max_tries = 3
    attempt = 0
    history = [{"role": h["role"], "content": h["content"]} for h in history]
    messages = [{"role": "system", "content": system_prompt}] + history + [{"role": "user", "content": message}]
    response = openai.chat.completions.create(model=model, messages=messages, tools=tools)

    while response.choices[0].finish_reason=="tool_calls":
                message = response.choices[0].message
                messages.append(message)
                for tool_call in message.tool_calls:
                    email = json.loads(tool_call.function.arguments).get("email")
                    record_email_tool(email)
                    messages.append({"role": "tool", "content": "Email recorded", "tool_call_id": tool_call.id})
                response = openai.chat.completions.create(model=model, messages=messages, tools=tools)
    

    while not evaluation_passed and attempt<max_tries:
        final_text = response.choices[0].message.content
        if not final_text:
                    break
        
        is_pass, evaluator_message = evaluate(response.choices[0].message.content)

        if is_pass:
                evaluation_passed = True
        else:
                attempt += 1
                messages.append({"role": "user", "content": f"Evaluator's Response: {evaluator_message}"})
                response = openai.chat.completions.create(model=model, messages=messages, tools=tools)
        
    

            
    return response.choices[0].message.content
````
